```python
# Append Items
mylist = ["Maha1","Maha2","Maha3","Maha4","Maha5","Maha6","Maha7"]
mylist.append("Python Developer and Gen AI Developer")
print(mylist)
# append means: add one item to the end of the list. not replace anything
```

```python
# Insert Items
mylist = ["Maha1","Maha2","Maha3","Maha4","Maha5","Maha6","Maha7"]
mylist.insert(3, "Python Developer and Gen AI Developer")
print(mylist)
```

```python
# Extend List
mylist = ["Maha1","Maha2","Maha3","Maha4","Maha5","Maha6","Maha7"]
mylist2 = ["Python Developer and Gen AI Developer","AI Engineer"]
mylist.extend(mylist2)
print(mylist)
```
