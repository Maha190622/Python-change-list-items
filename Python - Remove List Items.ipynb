```python
# Remove Specified Item
mylist = ["Maha1","Maha2","Maha3","Maha4","Maha5","Maha6","Maha7"]
mylist.remove("Maha3")
print(mylist)
```

```python
# Remove the first occurrence of "Maha2":
mylist = ["Maha1","Maha2","Maha3","Maha2","Maha5","Maha6","Maha7"]
mylist.remove("Maha2")
print(mylist)
```

```python
# Remove Specified Index
mylist = ["Maha1","Maha2","Maha3","Maha4","Maha5","Maha6","Maha7"]
removed_mylist = mylist.pop(4)
print(mylist)
```

```python
mylist.append(removed_mylist)
print(mylist)
```

```python
mylist.insert(4, removed_mylist)
print(mylist)
```

```python
# Clear the List
mylist = ["Maha1","Maha2","Maha3","Maha4","Maha5","Maha6","Maha7"]
mylist.clear()
print(mylist)
```
